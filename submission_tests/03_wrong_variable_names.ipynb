# Test 03 — Wrong variable names
Top comment: Uses different variable names (e.g., `epremium` instead of `equity_premium`) which may break rubric checks relying on names.

```python
import pandas as pd
from pathlib import Path
p = Path.cwd() / 'data' / 'GW05_original_monthly.csv'
df = pd.read_csv(p, sep=';', decimal=',')
df['date'] = pd.to_datetime(df['yyyymm'], format='%Y%m')
df.set_index('date', inplace=True)
# Construct but use different names
df['dp_index'] = df['D12']/df['Index']
df['epremium'] = df['CRSP_SPvw'] - df['Rfree']
# compute crude r2
from statsmodels.regression.linear_model import OLS
from statsmodels.tools import add_constant
x = df['dp_index'].shift(1).dropna()
y = df.loc[x.index, 'epremium']
reg = OLS(y, add_constant(x)).fit()
df_results = pd.DataFrame({'IS_R2_head': [reg.rsquared_adj*100]}, index=['dp_index'])
df_results.index.name = 'variable'
df_results
```
