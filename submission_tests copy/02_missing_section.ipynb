Top comment: This submission calculates in-sample regressions but omits any out-of-sample rolling procedure (missing required section).

```python
import numpy as np
import pandas as pd
from pathlib import Path
from statsmodels.regression.linear_model import OLS
from statsmodels.tools import add_constant
p = Path.cwd() / 'data' / 'GW05_original_monthly.csv'
data_monthly = pd.read_csv(p, sep=';', decimal=',')
data_monthly['date'] = pd.to_datetime(data_monthly['yyyymm'], format='%Y%m')
data_monthly.set_index('date', inplace=True)
data_monthly['dp'] = np.log(data_monthly['D12']) - np.log(data_monthly['Index'])
data_monthly['equity_premium'] = data_monthly['CRSP_SPvw'] - data_monthly['Rfree']
# In-sample only
x = data_monthly['dp'].shift(1).dropna()
y = data_monthly.loc[x.index, 'equity_premium']
reg = OLS(y, add_constant(x)).fit()
is_r2 = reg.rsquared_adj if hasattr(reg, 'rsquared_adj') else reg.rsquared
df_results = pd.DataFrame({'IS_R2_head': [is_r2*100]}, index=['dp'])
df_results.index.name = 'variable'
df_results
```
